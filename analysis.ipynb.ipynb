```python
import weather_module as wm
from weather_module import Station , Weatherstation 
import pandas as pd 

stations_df = wm.load_stations('stations.csv')
weather_df = wm.load_weather('weather.csv')

wm.check_missing_data(weather_df,'Weather')
print("-----------------------")
print("-----------------------")


######################################################

station_list = []
for i,row in stations_df.iterrows():
    obj = Station(
    row['Stations_ID'],
    row['Location'],
    row['geographic latitude'],
    row['geographic longitude'],
    row['height'],
    row['Operator']
    )
    station_list.append(obj)

weather_station_list = []
start_date = input('Enter the start date of the range for the weather station informations:') #2016
end_date = input('Enter the end date of the range for the weather station informations:') #2018

for st in station_list:
    ws_data = weather_df[weather_df['Stations_ID'] == st.Stations_ID]
    ws_data = weather_df[ (weather_df['Date']>= start_date) & (weather_df['Date']<= end_date)]
    ws_obj = Weatherstation(
        st.Stations_ID,
        st.location,
        st.latitude,
        st.longitude,
        st.height,
        st.operator,
        ws_data
    )
    weather_station_list.append(ws_obj)

######################################################

rows=[]
for ws in weather_station_list:
    rows.append({
        'Station ID':ws.Stations_ID,
        'Average Temp': ws.average_temperature(),
        'Average Wind': ws.average_windspeed(),
        'Average Sunshine': ws.average_sunshine()
    })
summery_df = pd.DataFrame(rows)
summery_df.head()

print("-----------------------")


target_id = int(input('Enter the Station ID:')) #880
selected_ws = None
for ws in weather_station_list:
    if ws.Stations_ID == target_id:
        selected_ws = ws
        break
if selected_ws is None:
    print('Station Not Found')
else:
    print('Found Station:',selected_ws.Stations_ID,selected_ws.location)
    print("-----------------------")
    selected_ws.plot_temperature()
    print("-----------------------")
    selected_ws.plot_windspeed()
    print("-----------------------")
    selected_ws.plot_sunshine()
    print("-----------------------")


# station_list[0].show_stationinfo()
# ws_data.head()

# wm.show_station_location(stations_df)


```
